```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import beta

# Parâmetros para diferentes distribuições Beta
params = [
    (0.5, 0.5),
    (2, 2),
    (2, 5),
    (5, 2),
    (1, 3),
    (3, 1),
    (5, 5),
    (10, 2)
]

x = np.linspace(0.001, 0.999, 500)  # evitar 0 e 1 exatos para evitar divisão por zero

# Plotando os gráficos
fig, axs = plt.subplots(4, 2, figsize=(14, 12))
axs = axs.flatten()

for i, (a, b) in enumerate(params):
    y = beta.pdf(x, a, b)
    axs[i].plot(x, y, label=f'α={a}, β={b}')
    axs[i].fill_between(x, y, alpha=0.3)
    axs[i].legend()
    axs[i].set_title(f'Distribuição Beta (α={a}, β={b})')
    axs[i].set_xlabel('x')
    axs[i].set_ylabel('Densidade')

plt.tight_layout()
plt.show()
```

Esses gráficos mostram como diferentes combinações de 
𝛼
α e 𝛽 moldam a distribuição Beta:


(α=0.5,β=0.5): U invertido — favorece extremos (0 e 1).


(α=2,β=2): Simétrica, concentrada no centro — neutra.

(α>β): Puxa para a direita (valores maiores).


(α<β): Puxa para a esquerda (valores menores).


(α=β): Simétrica em torno de 0.5.

Valores maiores de 𝛼 e β: Distribuições mais concentradas (menor variância).
